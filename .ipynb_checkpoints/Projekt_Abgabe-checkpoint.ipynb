### ML Projekt Jupyter Notebook Abgabe


```python
import pandas as pd
```

### 1. Daten laden

```python
train_df = pd.read_csv('data/project/train.csv',sep=';')

train_df
```

### 2. EDA durchführen

```python
import matplotlib.pyplot as plt

train_df["customerType"].value_counts().plot(kind="bar")
plt.xlabel("customerType")
plt.ylabel("Anzahl")
plt.title("Verteilung customerType")
plt.show()

train_df["totalAmount"].hist(bins=30)
plt.xlabel("totalAmount")
plt.ylabel("Anzahl")
plt.title("Histogram totalAmount")
plt.show()
```

```python
train_df["returnLabel"].value_counts().plot(kind="bar")
plt.xlabel("returnLabel")
plt.ylabel("Anzahl")
plt.title("Verteilung returnLabel (alle Daten)")
plt.show()

pd.crosstab(train_df["customerType"], train_df["returnLabel"]).plot(kind="bar", stacked=True)
plt.xlabel("customerType")
plt.ylabel("Anzahl")
plt.title("returnLabel nach customerType")
plt.show()
```

```python
train_df.boxplot(column="totalAmount", by="returnLabel")
plt.xlabel("returnLabel")
plt.ylabel("totalAmount")
plt.title("totalAmount nach returnLabel")
plt.suptitle("")  
plt.show()
```

### 3. Aufüllen fehlender Werte in den Trainingsdaten

```python
print(train_df[["customerType","totalAmount"]].isna().sum())

mode_ct = train_df["customerType"].mode(dropna=True)[0]
train_df["customerType"].fillna(mode_ct, inplace=True)

median_amt = train_df["totalAmount"].median()
train_df["totalAmount"].fillna(median_amt, inplace=True)

print(train_df[["customerType","totalAmount"]].isna().sum())
train_df
```

### 4. Transformieren kategorischer Features mittels One-hot-encoding

```python
cat_cols = train_df.select_dtypes(include=["object","category"]).columns.drop("returnLabel", errors="ignore")

train_df = pd.get_dummies(train_df, columns=["customerType"], drop_first=False) 
train_df
```

### 5. Features auf Basis des Attributs basket


```python
import ast

def parse_basket(x):
    if pd.isna(x):
        return []
    if isinstance(x, str):
        try:
            return list(ast.literal_eval(x))
        except Exception:
            return []
    return []

train_df["basket_list"] = train_df["basket"].apply(parse_basket)
```

```python
train_df["basket_total_items"] = train_df["basket_list"].apply(len)
train_df["basket_unique_items"] = train_df["basket_list"].apply(lambda L: len(set(L)))

exploded = train_df["basket_list"].explode()
K = 10
topk_items = exploded.value_counts().nlargest(K).index.tolist()

for itm in topk_items:
    col_name = f"basket_item_{itm}_count"
    train_df[col_name] = train_df["basket_list"].apply(lambda L, it=itm: L.count(it))
train_df
```

### 6. Skalieren der Features mit einem StandardScaler

```python
from sklearn.preprocessing import StandardScaler

drop_cols = ["returnLabel", "transactionId", "basket", "basket_list", "customerType"]

feature_cols = [c for c in train_df.columns if c not in drop_cols]

X = train_df[feature_cols]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(X_scaled, columns=feature_cols, index=train_df.index)

train_df_scaled = train_df.copy()
train_df_scaled[feature_cols] = X_scaled

train_df_scaled
```

### 7. Trainieren der Modelle


```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold

y = train_df["returnLabel"]

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

7.1. Logistische Regression

```python
log_reg = LogisticRegression(solver="liblinear", max_iter=1000)

param_grid_lr = {
    "C": [0.1, 1, 4, 5, 6, 10, 30, 100],
    "penalty": ["l1", "l2"]
}

grid_lr = GridSearchCV(
    estimator=log_reg,
    param_grid=param_grid_lr,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

grid_lr.fit(X, y)

print("Best params LR:", grid_lr.best_params_)
print("Best CV score LR:", grid_lr.best_score_)
```

7.2 Random Forest

```python
rf = RandomForestClassifier(random_state=42)

param_grid_rf = {
    "n_estimators": [60, 80, 100, 120, 140],
    "max_depth": [2, 4, 6]
}

grid_rf = GridSearchCV(
    estimator=rf,
    param_grid=param_grid_rf,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

grid_rf.fit(X, y)

print("Best params RF:", grid_rf.best_params_)
print("Best CV score RF:", grid_rf.best_score_)
```

7.3 Gradient Boosting Tree


```python
gb = GradientBoostingClassifier(random_state=42)

param_grid_gb = {
    "n_estimators": [60, 80, 100, 120, 140],
    "max_depth": [2, 4, 6]
}

grid_gb = GridSearchCV(
    estimator=gb,
    param_grid=param_grid_gb,
    cv=cv,
    scoring="accuracy",
    n_jobs=-1
)

grid_gb.fit(X, y)

print("Best params GB:", grid_gb.best_params_)
print("Best CV score GB:", grid_gb.best_score_)
```

### 8. Testdaten laden

```python
test_df = pd.read_csv('data/project/test.csv',sep=';')

test_df
```

### 9. Entfernen der Zeilen mit fehlenden Werten


```python
print(test_df.isna().sum())

test_df = test_df.dropna()

print(test_df[["customerType","totalAmount"]].isna().sum())
test_df
```

### 10. Transformieren der Attribute wie bei Trainingsdaten

```python
cat_cols = test_df.select_dtypes(include=["object","category"]).columns.drop("returnLabel", errors="ignore")

test_df = pd.get_dummies(test_df, columns=["customerType"], drop_first=False) 
test_df
```

```python
test_df["basket_list"] = test_df["basket"].apply(parse_basket)
test_df["basket_total_items"] = test_df["basket_list"].apply(len)
test_df["basket_unique_items"] = test_df["basket_list"].apply(lambda L: len(set(L)))

exploded = test_df["basket_list"].explode()
K = 10
topk_items = exploded.value_counts().nlargest(K).index.tolist()

for itm in topk_items:
    col_name = f"basket_item_{itm}_count"
    test_df[col_name] = test_df["basket_list"].apply(lambda L, it=itm: L.count(it))
test_df
```

### 11. Skalieren der Daten im gleichen Maße wie beim Training

```python
drop_cols = ["returnLabel", "transactionId", "basket", "basket_list", "customerType"]

feature_cols = [c for c in test_df.columns if c not in drop_cols]

X_test = test_df[feature_cols]

X_scaled_test = scaler.fit_transform(X_test)

X_scaled_test = pd.DataFrame(X_scaled_test, columns=feature_cols, index=test_df.index)

test_df_scaled = test_df.copy()
test_df_scaled[feature_cols] = X_scaled_test

test_df_scaled
```

### 12. Vorhersage der Testdaten mit allen Modellen

```python
drop_cols = ["returnLabel", "transactionId", "basket", "basket_list"]

feature_cols = [c for c in train_df_scaled.columns if c not in drop_cols]

X_train = train_df_scaled[feature_cols]
y_train = train_df_scaled["returnLabel"]

X_test = test_df_scaled[feature_cols]
y_test = test_df_scaled["returnLabel"]
```

```python
best_lr = grid_lr.best_estimator_
best_rf = grid_rf.best_estimator_
best_gb = grid_gb.best_estimator_

y_pred_lr = best_lr.predict(X_test)
y_pred_rf = best_rf.predict(X_test)
y_pred_gb = best_gb.predict(X_test)
```

### 13. Berechnen von Accuracy, Precision und Recall

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score

def eval_model(name, y_true, y_pred):
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, pos_label=1)
    rec = recall_score(y_true, y_pred, pos_label=1)
    print(f"{name}:")
    print(f"  Accuracy : {acc:.4f}")
    print(f"  Precision: {prec:.4f}")
    print(f"  Recall   : {rec:.4f}")
    print()

print("== Testdaten ==")
eval_model("Logistische Regression", y_test, y_pred_lr)
eval_model("Random Forest", y_test, y_pred_rf)
eval_model("Gradient Boosting", y_test, y_pred_gb)
```

### 14. Accuracy auf den Trainingsdaten

```python
y_train_pred_lr = best_lr.predict(X_train)
y_train_pred_rf = best_rf.predict(X_train)
y_train_pred_gb = best_gb.predict(X_train)

def train_test_compare(name, y_tr, y_tr_pred, y_te, y_te_pred):
    acc_tr = accuracy_score(y_tr, y_tr_pred)
    acc_te = accuracy_score(y_te, y_te_pred)
    diff = acc_tr - acc_te
    print(f"{name}:")
    print(f"  Train-Accuracy: {acc_tr:.4f}")
    print(f"  Test-Accuracy : {acc_te:.4f}")
    print(f"  Differenz     : {diff:.4f}")
    if diff > 0.05:
        print("  → Overfitting (Unterschied > 5%)")
    else:
        print("  → Kein starkes Overfitting (Unterschied ≤ 5%)")
    print()

print("== Vergleich Train vs. Test ==")
train_test_compare("Logistische Regression", y_train, y_train_pred_lr, y_test, y_pred_lr)
train_test_compare("Random Forest",        y_train, y_train_pred_rf, y_test, y_pred_rf)
train_test_compare("Gradient Boosting",    y_train, y_train_pred_gb, y_test, y_pred_gb)
```

### 15. Falsch klassifizierte Datenpunkte
15.1. Indizes der falsch klassifizierten Datenpunkte

```python
import numpy as np

wrong_lr = y_pred_lr != y_test
wrong_rf = y_pred_rf != y_test
wrong_gb = y_pred_gb != y_test

idx_wrong_lr = set(X_test.index[wrong_lr])
idx_wrong_rf = set(X_test.index[wrong_rf])
idx_wrong_gb = set(X_test.index[wrong_gb])

print("Anzahl Fehler LR:", len(idx_wrong_lr))
print("Anzahl Fehler RF:", len(idx_wrong_rf))
print("Anzahl Fehler GB:", len(idx_wrong_gb))
```

15.2. Anzahl der Datenpunkte, die alle Modelle falsch hatten

```python

common_wrong = idx_wrong_lr & idx_wrong_rf & idx_wrong_gb

print("Anzahl Datenpunkte, die alle drei falsch klassifizieren:", len(common_wrong))
print("Indizes:", list(common_wrong)[:20])
```

### 16. Feature Importance der drei Modelle vergleichen

```python
lr_importances = np.abs(best_lr.coef_[0])

rf_importances = best_rf.feature_importances_

gb_importances = best_gb.feature_importances_

fi_df = pd.DataFrame({
    "feature": feature_cols,
    "LR": lr_importances,
    "RF": rf_importances,
    "GB": gb_importances
})

fi_df["mean_importance"] = fi_df[["LR", "RF", "GB"]].mean(axis=1)
fi_df = fi_df.sort_values("mean_importance", ascending=False)

top_n = 15
fi_top = fi_df.head(top_n)

x = np.arange(len(fi_top)) 
width = 0.25

plt.figure(figsize=(12, 6))
plt.bar(x - width, fi_top["LR"], width=width, label="Logistic Regression")
plt.bar(x,         fi_top["RF"], width=width, label="Random Forest")
plt.bar(x + width, fi_top["GB"], width=width, label="Gradient Boosting")

plt.xticks(x, fi_top["feature"], rotation=90)
plt.ylabel("Feature Importance (|coef| bzw. importance)")
plt.title("Feature Importances – Vergleich der drei Modelle")
plt.legend()
plt.tight_layout()
plt.show()
```
